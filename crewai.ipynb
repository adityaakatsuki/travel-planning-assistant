```python
import os
from crewai import Agent,Task,Crew,Process,LLM
from crewai_tools import SerperDevTool
```

```python
from google.colab import userdata
serper_api=userdata.get('serper_key')
gemini_key=userdata.get('gemini_Aikey')
os.environ["GOOGLE_API_KEY"]=gemini_key
os.environ["SERPER_API_KEY"]=serper_api
```

step3 intialize the tools

```python
search_tool=SerperDevTool()
#define the ai model
llm=LLM(model="gemini/gemini-3.5-flash")
verbose=True
temperature=0.5
llm
```

```python
# Travel researcher Agent (finds historical sites and weather)
researcher=Agent(
    role="Travel researcher",
    goal="""find historical sites ,public transport hotels,and real time weather information for {destination}""",
    verbose=True,
    memory=True,
    backstory="you are an expert travel researcher ",
    llm=llm,
    tools=[search_tool],
    allow_delegation=True

)

# Budget Planner  Agent (ensures trip stays under $1500)
budget_planner=Agent(
    role="Budget_planner",
    goal="""find budget flights,hotels,and activities with in {budget} for {destination}""",
    verbose=True,
    memory=True,
    backstory="you are an skilled budget analyst ensuring trip fit with financial  ",
    llm=llm,
    tools=[search_tool],
    allow_delegation=False
)

# Itenary planner Agent (creates a balanced 3 day plan)
itinerary_planner=Agent(
    role="itienerary planner",
    goal="""create a 3 day itienary  for {destination}""",
    verbose=True,
    memory=True,
    backstory="you are an expert in trip planning ",
    llm=llm,
    tools=[search_tool],
    allow_delegation=False

)
```

```python
# Travel researcher Agent (finds historical sites and weather)
researcher=Agent(
    role="Travel researcher",
    goal="""find historical sites ,public transport hotels,and real time weather information for {destination}""",
    verbose=True,
    memory=True,
    backstory="you are an expert travel researcher ",
    llm=llm,
    tools=[search_tool],
    allow_delegation=True
)

#Travel researcher Task
researcher_task=Task(
    description="find historical sites with weather forecasts and public transport hotels for {destination}",
    expected_output="list of historical sites,a real time weather update and 3 hotel options nearby public transport",
    tools=[search_tool],
    agent=researcher
)
#Budget Estimation Task
budget_task=Task(
    description="find budget flights hotel options and daily food/transport and {destination} ensure total cost is in {budget}",
    expected_output="a full cost breakdown ensuring budget $1500 is maintained ",
    tools=[search_tool],
    agent=budget_planner
)
#Itienary planning Task
itienary_task=Task(
    description="plan a 3 day itienary  for {destination} focussing on history",
    expected_output="A detailed 3 day plan considering weather and budget with transport recommendations ",
    tools=[search_tool],
    agent=itinerary_planner
)
```

```python
# Crew setup all Agents working together
crew=Crew(
    agents=[researcher,budget_planner,itinerary_planner],
    tasks=[researcher_task,budget_task,itienary_task],
    process=Process.sequential
)
crew
```

```python
result= await crew.kickoff_async(inputs={'destination':'London','budget':'$1500'})
result
```
